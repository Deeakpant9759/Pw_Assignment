```python
# Question 24. Write a Pandas script to find the mean of a specific column in a DataFrame.
df = sns.load_dataset('titanic')
df['age'].mean()
df.head()
```

```python
# Question 25. Create a scatter plot using Matplotlib.
plt.scatter(df['age'],df['fare'])
plt.title('age vs fare')
plt.xlabel('age')
plt.ylabel('fare')

```

```python
# Question 26. Calculate the correlation matrix using Seaborn and visualize it with a heatmap
plt.figure(figsize=(10, 8))
edf = pd.DataFrame({
    'A':[1,2,3,4,5],
    'B':[5,4,3,2,1],
    'C':[2,3,4,5,6]
})

corr_matrix = edf.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0,
            fmt='.2f', square=True, linewidths=1)
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
# Question27. Generate a bar plot using Plotly.
px.bar(df,x='sex',y='age')
```

```python
# Question 28. Create a DataFrame and add a new column based on an existing column.
df['new_col'] = df['age'] + df['fare']
df.head()
```

```python
# Question 29. Perform element-wise multiplication of two NumPy arrays.
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([5, 4, 3, 2, 1])
arr1 * arr2
```

```python
# Question 30. Create a line plot with multiple lines using Matplotlib.
df["age_group"] = pd.cut(df["age"], bins=10)
survival_rates =df.groupby(["age_group", "sex"])["survived"].mean().unstack()
plt.plot(survival_rates.index.astype(str), survival_rates["male"], label="Male")
plt.plot(survival_rates.index.astype(str), survival_rates["female"], label="Female")
plt.legend()
plt.xticks(rotation=45)
plt.xlabel("Age Groups")
plt.ylabel("Survival Rate")
plt.title("Titanic Survival Rates by Age Group and Gender")
plt.show()
```
